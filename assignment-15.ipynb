```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import math
import numpy as np
from transformers import BertTokenizer, BertForSequenceClassification, BertForQuestionAnswering
from transformers import get_linear_schedule_with_warmup
from datasets import load_dataset
from tqdm import tqdm
import time
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, f1_score, classification_report
import pandas as pd
import warnings
warnings.filterwarnings('ignore')



class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads, dropout=0.1):
        super().__init__()
        assert d_model % num_heads == 0
        
        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads
        
        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)
        self.W_o = nn.Linear(d_model, d_model)
        
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, query, key, value, mask=None):
        batch_size = query.size(0)
        
        Q = self.W_q(query).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        K = self.W_k(key).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        V = self.W_v(value).view(batch_size, -1, self.num_heads, self.d_k).transpose(1, 2)
        
        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)
        
        if mask is not None:
            scores = scores.masked_fill(mask == 0, -1e9)
        
        attn = F.softmax(scores, dim=-1)
        attn = self.dropout(attn)
        
        context = torch.matmul(attn, V)
        context = context.transpose(1, 2).contiguous().view(batch_size, -1, self.d_model)
        
        return self.W_o(context)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super().__init__()
        self.linear1 = nn.Linear(d_model, d_ff)
        self.linear2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, x):
        return self.linear2(self.dropout(F.gelu(self.linear1(x))))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super().__init__()
        self.attention = MultiHeadAttention(d_model, num_heads, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.ff = FeedForward(d_model, d_ff, dropout)
        self.norm2 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, x, mask=None):
        attn_out = self.attention(x, x, x, mask)
        x = self.norm1(x + self.dropout(attn_out))
        ff_out = self.ff(x)
        x = self.norm2(x + self.dropout(ff_out))
        return x


class BERTEmbedding(nn.Module):
    def __init__(self, vocab_size, d_model, max_len=512, dropout=0.1):
        super().__init__()
        self.token_embedding = nn.Embedding(vocab_size, d_model)
        self.position_embedding = nn.Embedding(max_len, d_model)
        self.segment_embedding = nn.Embedding(2, d_model)
        
        self.dropout = nn.Dropout(dropout)
        self.norm = nn.LayerNorm(d_model)
        
    def forward(self, x, segment_ids=None):
        batch_size, seq_len = x.size()
        
        token_emb = self.token_embedding(x)
        positions = torch.arange(seq_len, device=x.device).unsqueeze(0).expand(batch_size, seq_len)
        position_emb = self.position_embedding(positions)
        
        if segment_ids is None:
            segment_ids = torch.zeros_like(x)
        segment_emb = self.segment_embedding(segment_ids)
        
        embeddings = token_emb + position_emb + segment_emb
        return self.dropout(self.norm(embeddings))


class BERT(nn.Module):
    def __init__(self, vocab_size, d_model=768, num_layers=12, num_heads=12, 
                 d_ff=3072, max_len=512, dropout=0.1):
        super().__init__()
        self.d_model = d_model
        
        self.embedding = BERTEmbedding(vocab_size, d_model, max_len, dropout)
        self.encoder_layers = nn.ModuleList([
            TransformerBlock(d_model, num_heads, d_ff, dropout)
            for _ in range(num_layers)
        ])
        
    def forward(self, x, segment_ids=None, attention_mask=None):
        if attention_mask is not None:
            attention_mask = attention_mask.unsqueeze(1).unsqueeze(2)
        
        x = self.embedding(x, segment_ids)
        
        for layer in self.encoder_layers:
            x = layer(x, attention_mask)
        
        return x


class CustomBERTForSequenceClassification(nn.Module):
    def __init__(self, bert_model, num_classes=2):
        super().__init__()
        self.bert = bert_model
        self.dropout = nn.Dropout(0.1)
        self.classifier = nn.Linear(bert_model.d_model, num_classes)
        
    def forward(self, input_ids, segment_ids=None, attention_mask=None):
        sequence_output = self.bert(input_ids, segment_ids, attention_mask)
        cls_output = sequence_output[:, 0, :]
        cls_output = self.dropout(cls_output)
        logits = self.classifier(cls_output)
        return logits


class CustomBERTForQuestionAnswering(nn.Module):
    def __init__(self, bert_model):
        super().__init__()
        self.bert = bert_model
        self.qa_outputs = nn.Linear(bert_model.d_model, 2)
        
    def forward(self, input_ids, segment_ids=None, attention_mask=None):
        sequence_output = self.bert(input_ids, segment_ids, attention_mask)
        logits = self.qa_outputs(sequence_output)
        start_logits, end_logits = logits.split(1, dim=-1)
        start_logits = start_logits.squeeze(-1)
        end_logits = end_logits.squeeze(-1)
        return start_logits, end_logits


# DATASET CLASSES

class SentimentDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
        
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = self.texts[idx]
        label = self.labels[idx]
        
        encoding = self.tokenizer(text, max_length=self.max_len, padding='max_length',
                                  truncation=True, return_tensors='pt')
        
        return {
            'input_ids': encoding['input_ids'].squeeze(),
            'attention_mask': encoding['attention_mask'].squeeze(),
            'labels': torch.tensor(label, dtype=torch.long)
        }


class SQuADDataset(Dataset):
    def __init__(self, data, tokenizer, max_len=384):
        self.data = data
        self.tokenizer = tokenizer
        self.max_len = max_len
        
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        item = self.data[idx]
        question = item['question']
        context = item['context']
        answer = item['answers']['text'][0] if len(item['answers']['text']) > 0 else ""
        answer_start = item['answers']['answer_start'][0] if len(item['answers']['answer_start']) > 0 else 0
        
        encoding = self.tokenizer(
            question,
            context,
            max_length=self.max_len,
            padding='max_length',
            truncation='only_second',
            return_tensors='pt',
            return_offsets_mapping=True
        )
        
        input_ids = encoding['input_ids'].squeeze()
        attention_mask = encoding['attention_mask'].squeeze()
        token_type_ids = encoding['token_type_ids'].squeeze() if 'token_type_ids' in encoding else torch.zeros_like(input_ids)
        offset_mapping = encoding['offset_mapping'].squeeze()
        
        start_position = 0
        end_position = 0
        
        if answer:
            answer_end = answer_start + len(answer)
            for idx, (offset_start, offset_end) in enumerate(offset_mapping):
                if offset_start <= answer_start < offset_end:
                    start_position = idx
                if offset_start < answer_end <= offset_end:
                    end_position = idx
                    break
        
        return {
            'input_ids': input_ids,
            'attention_mask': attention_mask,
            'token_type_ids': token_type_ids,
            'start_positions': torch.tensor(start_position, dtype=torch.long),
            'end_positions': torch.tensor(end_position, dtype=torch.long)
        }


# ===========================================================================
# TRAINING AND EVALUATION FUNCTIONS

def train_sentiment_model(model, train_loader, val_loader, epochs, lr, device, model_name):
    """Train sentiment classification model and return metrics"""
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()
    
    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss': [], 'val_acc': [],
        'train_time': [], 'inference_time': []
    }
    
    best_val_acc = 0
    
    for epoch in range(epochs):
        # Training
        model.train()
        total_loss = 0
        correct = 0
        total = 0
        start_time = time.time()
        
        progress_bar = tqdm(train_loader, desc=f'{model_name} - Epoch {epoch+1}/{epochs}')
        for batch in progress_bar:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)
            
            optimizer.zero_grad()
            
            # Handle different model outputs
            if isinstance(model, BertForSequenceClassification):
                outputs = model(input_ids, attention_mask=attention_mask)
                logits = outputs.logits
            else:
                logits = model(input_ids, attention_mask=attention_mask)
            
            loss = criterion(logits, labels)
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            _, predicted = torch.max(logits, 1)
            correct += (predicted == labels).sum().item()
            total += labels.size(0)
            
            progress_bar.set_postfix({'loss': loss.item(), 'acc': correct/total})
        
        train_time = time.time() - start_time
        train_acc = correct / total
        train_loss = total_loss / len(train_loader)
        
        # Validation
        model.eval()
        val_loss = 0
        val_correct = 0
        val_total = 0
        all_preds = []
        all_labels = []
        
        start_time = time.time()
        with torch.no_grad():
            for batch in val_loader:
                input_ids = batch['input_ids'].to(device)
                attention_mask = batch['attention_mask'].to(device)
                labels = batch['labels'].to(device)
                
                if isinstance(model, BertForSequenceClassification):
                    outputs = model(input_ids, attention_mask=attention_mask)
                    logits = outputs.logits
                else:
                    logits = model(input_ids, attention_mask=attention_mask)
                
                loss = criterion(logits, labels)
                val_loss += loss.item()
                
                _, predicted = torch.max(logits, 1)
                val_correct += (predicted == labels).sum().item()
                val_total += labels.size(0)
                
                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        
        inference_time = time.time() - start_time
        val_acc = val_correct / val_total
        val_loss = val_loss / len(val_loader)
        
        # Store metrics
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['train_time'].append(train_time)
        history['inference_time'].append(inference_time)
        
        print(f'\nEpoch {epoch+1}:')
        print(f'  Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.4f}')
        print(f'  Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.4f}')
        print(f'  Train Time: {train_time:.2f}s, Inference Time: {inference_time:.2f}s')
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
    
    # Final metrics
    f1 = f1_score(all_labels, all_preds, average='weighted')
    history['best_val_acc'] = best_val_acc
    history['final_f1'] = f1
    
    return history


def train_qa_model(model, train_loader, val_loader, epochs, lr, device, model_name):
    """Train question answering model and return metrics"""
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    
    history = {
        'train_loss': [], 'val_loss': [],
        'train_time': [], 'inference_time': []
    }
    
    for epoch in range(epochs):
        model.train()
        total_loss = 0
        start_time = time.time()
        
        progress_bar = tqdm(train_loader, desc=f'{model_name} - Epoch {epoch+1}/{epochs}')
        for batch in progress_bar:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            token_type_ids = batch['token_type_ids'].to(device)
            start_positions = batch['start_positions'].to(device)
            end_positions = batch['end_positions'].to(device)
            
            optimizer.zero_grad()
            
            # Handle different model outputs
            if isinstance(model, BertForQuestionAnswering):
                outputs = model(input_ids, token_type_ids=token_type_ids, attention_mask=attention_mask,
                               start_positions=start_positions, end_positions=end_positions)
                loss = outputs.loss
            else:
                start_logits, end_logits = model(input_ids, token_type_ids, attention_mask)
                loss_fct = nn.CrossEntropyLoss()
                start_loss = loss_fct(start_logits, start_positions)
                end_loss = loss_fct(end_logits, end_positions)
                loss = (start_loss + end_loss) / 2
            
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            progress_bar.set_postfix({'loss': loss.item()})
        
        train_time = time.time() - start_time
        train_loss = total_loss / len(train_loader)
        
        # Validation
        model.eval()
        val_loss = 0
        start_time = time.time()
        
        with torch.no_grad():
            for batch in val_loader:
                input_ids = batch['input_ids'].to(device)
                attention_mask = batch['attention_mask'].to(device)
                token_type_ids = batch['token_type_ids'].to(device)
                start_positions = batch['start_positions'].to(device)
                end_positions = batch['end_positions'].to(device)
                
                if isinstance(model, BertForQuestionAnswering):
                    outputs = model(input_ids, token_type_ids=token_type_ids, attention_mask=attention_mask,
                                   start_positions=start_positions, end_positions=end_positions)
                    loss = outputs.loss
                else:
                    start_logits, end_logits = model(input_ids, token_type_ids, attention_mask)
                    loss_fct = nn.CrossEntropyLoss()
                    start_loss = loss_fct(start_logits, start_positions)
                    end_loss = loss_fct(end_logits, end_positions)
                    loss = (start_loss + end_loss) / 2
                
                val_loss += loss.item()
        
        inference_time = time.time() - start_time
        val_loss = val_loss / len(val_loader)
        
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['train_time'].append(train_time)
        history['inference_time'].append(inference_time)
        
        print(f'\nEpoch {epoch+1}:')
        print(f'  Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')
        print(f'  Train Time: {train_time:.2f}s, Inference Time: {inference_time:.2f}s')
    
    return history


def count_parameters(model):
    """Count trainable parameters in model"""
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


# ============================================================================
# COMPARISON AND EXPERIMENTATION
# ============================================================================

def run_sentiment_comparison(device='cuda'):
    """Compare different BERT configurations on sentiment analysis"""
    
    print("\n" + "="*80)
    print("SENTIMENT CLASSIFICATION COMPARISON")
    print("="*80)
    
    # Load tokenizer
    tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
    vocab_size = len(tokenizer)
    
    # Load IMDb dataset (or use your own)
    print("\nLoading IMDb dataset...")
    try:
        dataset = load_dataset('imdb', split='train[:5000]')  # Use subset for faster training
        train_data = dataset.train_test_split(test_size=0.2, seed=42)
        
        train_texts = train_data['train']['text']
        train_labels = train_data['train']['label']
        val_texts = train_data['test']['text']
        val_labels = train_data['test']['label']
    except:
        # Fallback to dummy data
        print("Using dummy data for demonstration...")
        train_texts = ["This movie is great!"] * 100 + ["This movie is terrible!"] * 100
        train_labels = [1] * 100 + [0] * 100
        val_texts = ["Amazing film!"] * 20 + ["Awful movie!"] * 20
        val_labels = [1] * 20 + [0] * 20
    
    # Create datasets
    train_dataset = SentimentDataset(train_texts, train_labels, tokenizer, max_len=128)
    val_dataset = SentimentDataset(val_texts, val_labels, tokenizer, max_len=128)
    
    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=16)
    
    results = {}
    
    # Configuration 1: Small Custom BERT
    print("\n" + "-"*80)
    print("Configuration 1: Small Custom BERT (128-dim, 2 layers)")
    print("-"*80)
    bert_small = BERT(vocab_size=vocab_size, d_model=128, num_layers=2, num_heads=4, 
                      d_ff=512, max_len=128, dropout=0.1)
    model_small = CustomBERTForSequenceClassification(bert_small, num_classes=2)
    print(f"Parameters: {count_parameters(model_small):,}")
    
    results['Small Custom (128d, 2L)'] = train_sentiment_model(
        model_small, train_loader, val_loader, epochs=3, lr=2e-4, device=device,
        model_name="Small Custom"
    )
    
    # Configuration 2: Medium Custom BERT
    print("\n" + "-"*80)
    print("Configuration 2: Medium Custom BERT (256-dim, 4 layers)")
    print("-"*80)
    bert_medium = BERT(vocab_size=vocab_size, d_model=256, num_layers=4, num_heads=8, 
                       d_ff=1024, max_len=128, dropout=0.1)
    model_medium = CustomBERTForSequenceClassification(bert_medium, num_classes=2)
    print(f"Parameters: {count_parameters(model_medium):,}")
    
    results['Medium Custom (256d, 4L)'] = train_sentiment_model(
        model_medium, train_loader, val_loader, epochs=3, lr=2e-4, device=device,
        model_name="Medium Custom"
    )
    
    # Configuration 3: Large Custom BERT
    print("\n" + "-"*80)
    print("Configuration 3: Large Custom BERT (512-dim, 6 layers)")
    print("-"*80)
    bert_large = BERT(vocab_size=vocab_size, d_model=512, num_layers=6, num_heads=8, 
                      d_ff=2048, max_len=128, dropout=0.1)
    model_large = CustomBERTForSequenceClassification(bert_large, num_classes=2)
    print(f"Parameters: {count_parameters(model_large):,}")
    
    results['Large Custom (512d, 6L)'] = train_sentiment_model(
        model_large, train_loader, val_loader, epochs=3, lr=2e-4, device=device,
        model_name="Large Custom"
    )
    
    # Configuration 4: Pre-trained BERT
    print("\n" + "-"*80)
    print("Configuration 4: Pre-trained BERT-base (768-dim, 12 layers)")
    print("-"*80)
    model_pretrained = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=2)
    print(f"Parameters: {count_parameters(model_pretrained):,}")
    
    results['Pre-trained BERT'] = train_sentiment_model(
        model_pretrained, train_loader, val_loader, epochs=3, lr=2e-5, device=device,
        model_name="Pre-trained BERT"
    )
    
    return results


def run_qa_comparison(device='cuda'):
    """Compare different BERT configurations on question answering"""
    
    print("\n" + "="*80)
    print("QUESTION ANSWERING COMPARISON (SQuAD)")
    print("="*80)
    
    tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
    vocab_size = len(tokenizer)
    
    # Load SQuAD dataset
    print("\nLoading SQuAD dataset...")
    try:
        dataset = load_dataset('squad', split='train[:1000]')
        train_data = dataset.train_test_split(test_size=0.2, seed=42)
        
        train_dataset = SQuADDataset(train_data['train'], tokenizer, max_len=384)
        val_dataset = SQuADDataset(train_data['test'], tokenizer, max_len=384)
    except:
        print("Could not load SQuAD dataset. Skipping QA comparison.")
        return {}
    
    train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=8)
    
    results = {}
    
    # Custom BERT configurations
    print("\n" + "-"*80)
    print("Configuration 1: Medium Custom BERT (256-dim, 4 layers)")
    print("-"*80)
    bert_medium = BERT(vocab_size=vocab_size, d_model=256, num_layers=4, num_heads=8, 
                       d_ff=1024, max_len=384, dropout=0.1)
    model_medium = CustomBERTForQuestionAnswering(bert_medium)
    print(f"Parameters: {count_parameters(model_medium):,}")
    
    results['Medium Custom (256d, 4L)'] = train_qa_model(
        model_medium, train_loader, val_loader, epochs=2, lr=3e-4, device=device,
        model_name="Medium Custom"
    )
    
    # Pre-trained BERT
    print("\n" + "-"*80)
    print("Configuration 2: Pre-trained BERT-base")
    print("-"*80)
    model_pretrained = BertForQuestionAnswering.from_pretrained('bert-base-uncased')
    print(f"Parameters: {count_parameters(model_pretrained):,}")
    
    results['Pre-trained BERT'] = train_qa_model(
        model_pretrained, train_loader, val_loader, epochs=2, lr=3e-5, device=device,
        model_name="Pre-trained BERT"
    )
    
    return results


def visualize_results(sentiment_results, qa_results=None):
    """Visualize comparison results"""
    
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    fig.suptitle('BERT Model Comparison', fontsize=16, fontweight='bold')
    
    # Extract data
    models = list(sentiment_results.keys())
    colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4']
    
    # 1. Parameters comparison (extract from results if available)
    # Create dummy params based on model names for visualization
    param_map = {
        'Small Custom (128d, 2L)': 0.5,
        'Medium Custom (256d, 4L)': 4.2,
        'Large Custom (512d, 6L)': 25.0,
        'Pre-trained BERT': 110.0
    }
    params = [param_map.get(m, 1.0) for m in models]
    
    axes[0, 0].bar(range(len(models)), params, color=colors[:len(models)])
    axes[0, 0].set_xticks(range(len(models)))
    axes[0, 0].set_xticklabels(models, rotation=45, ha='right', fontsize=9)
    axes[0, 0].set_ylabel('Parameters (Millions)')
    axes[0, 0].set_title('Model Size (Parameters)')
    axes[0, 0].grid(axis='y', alpha=0.3)
    
    # 2. Validation accuracy
    val_accs = [sentiment_results[m]['best_val_acc'] for m in models]
    axes[0, 1].bar(range(len(models)), val_accs, color=colors[:len(models)])
    axes[0, 1].set_xticks(range(len(models)))
    axes[0, 1].set_xticklabels(models, rotation=45, ha='right')
    axes[0, 1].set_ylabel('Accuracy')
    axes[0, 1].set_title('Best Validation Accuracy (Sentiment)')
    axes[0, 1].set_ylim([0, 1])
    axes[0, 1].grid(axis='y', alpha=0.3)
    
    # 3. Training time
    train_times = [sum(sentiment_results[m]['train_time']) for m in models]
    axes[0, 2].bar(range(len(models)), train_times, color=colors[:len(models)])
    axes[0, 2].set_xticks(range(len(models)))
    axes[0, 2].set_xticklabels(models, rotation=45, ha='right')
    axes[0, 2].set_ylabel('Time (seconds)')
    axes[0, 2].set_title('Total Training Time')
    axes[0, 2].grid(axis='y', alpha=0.3)
    
    # 4. Learning curves
    for i, model in enumerate(models):
        epochs = range(1, len(sentiment_results[model]['train_acc']) + 1)
        axes[1, 0].plot(epochs, sentiment_results[model]['train_acc'], 
                       marker='o', label=f'{model} (Train)', color=colors[i], linestyle='-')
        axes[1, 0].plot(epochs, sentiment_results[model]['val_acc'], 
                       marker='s', label=f'{model} (Val)', color=colors[i], linestyle='--')
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('Accuracy')
    axes[1, 0].set_title('Learning Curves (Sentiment)')
    axes[1, 0].legend(fontsize=8)
    axes[1, 0].grid(alpha=0.3)
    
    # 5. Inference time
    inference_times = [sentiment_results[m]['inference_time'][-1] for m in models]
    axes[1, 1].bar(range(len(models)), inference_times, color=colors[:len(models)])
    axes[1, 1].set_xticks(range(len(models)))
    axes[1, 1].set_xticklabels(models, rotation=45, ha='right')
    axes[1, 1].set_ylabel('Time (seconds)')
    axes[1, 1].set_title('Inference Time per Epoch')
    axes[1, 1].grid(axis='y', alpha=0.3)
    
    # 6. Summary table
    axes[1, 2].axis('off')
    summary_data = []
    for model in models:
        summary_data.append([
            model,
            f"{sentiment_results[model]['best_val_acc']:.4f}",
            f"{sentiment_results[model]['final_f1']:.4f}",
            f"{sum(sentiment_results[model]['train_time']):.1f}s"
        ])
    
    table = axes[1, 2].table(cellText=summary_data,
                            colLabels=['Model', 'Best Acc', 'F1 Score', 'Train Time'],
                            cellLoc='left',
                            loc='center',
                            bbox=[0, 0, 1, 1])
    table.auto_set_font_size(False)
    table.set_fontsize(9)
    table.scale(1, 2)
    axes[1, 2].set_title('Performance Summary')
    
    # Style table
    for i in range(len(models) + 1):
        for j in range(4):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#E8E8E8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F8F8F8')
    
    plt.tight_layout()
    plt.savefig('bert_comparison_results.png', dpi=300, bbox_inches='tight')
    print("\n✓ Results saved to 'bert_comparison_results.png'")
    plt.show()


def print_detailed_comparison(sentiment_results, qa_results=None):
    """Print detailed comparison table"""
    
    print("\n" + "="*100)
    print("DETAILED PERFORMANCE COMPARISON")
    print("="*100)
    
    # Sentiment Analysis Results
    print("\n--- SENTIMENT CLASSIFICATION RESULTS ---\n")
    
    data = []
    for model_name, metrics in sentiment_results.items():
        data.append({
            'Model': model_name,
            'Best Val Acc': f"{metrics['best_val_acc']:.4f}",
            'Final F1': f"{metrics['final_f1']:.4f}",
            'Avg Train Loss': f"{np.mean(metrics['train_loss']):.4f}",
            'Avg Val Loss': f"{np.mean(metrics['val_loss']):.4f}",
            'Total Train Time': f"{sum(metrics['train_time']):.2f}s",
            'Avg Inference Time': f"{np.mean(metrics['inference_time']):.2f}s"
        })
    
    df = pd.DataFrame(data)
    print(df.to_string(index=False))
    
    # Question Answering Results
    if qa_results:
        print("\n--- QUESTION ANSWERING RESULTS ---\n")
        
        qa_data = []
        for model_name, metrics in qa_results.items():
            qa_data.append({
                'Model': model_name,
                'Final Train Loss': f"{metrics['train_loss'][-1]:.4f}",
                'Final Val Loss': f"{metrics['val_loss'][-1]:.4f}",
                'Total Train Time': f"{sum(metrics['train_time']):.2f}s",
                'Avg Inference Time': f"{np.mean(metrics['inference_time']):.2f}s"
            })
        
        df_qa = pd.DataFrame(qa_data)
        print(df_qa.to_string(index=False))
    
    print("\n" + "="*100)
    
    # Key Insights
    print("\nKEY INSIGHTS:")
    print("-" * 100)
    
    best_acc_model = max(sentiment_results.items(), key=lambda x: x[1]['best_val_acc'])
    fastest_model = min(sentiment_results.items(), key=lambda x: sum(x[1]['train_time']))
    
    print(f"• Best Accuracy: {best_acc_model[0]} with {best_acc_model[1]['best_val_acc']:.4f}")
    print(f"• Fastest Training: {fastest_model[0]} with {sum(fastest_model[1]['train_time']):.2f}s total")
    print(f"• Pre-trained BERT advantage: Shows transfer learning benefits")
    print(f"• Custom BERT: Lighter models train faster but may need more data/epochs")
    
    # Calculate efficiency score (accuracy per training time)
    print("\n• Efficiency Ranking (Accuracy/Training Time):")
    efficiency = {name: metrics['best_val_acc'] / sum(metrics['train_time']) 
                  for name, metrics in sentiment_results.items()}
    for i, (name, score) in enumerate(sorted(efficiency.items(), key=lambda x: x[1], reverse=True), 1):
        print(f"  {i}. {name}: {score:.6f}")


def hyperparameter_experiment(device='cuda'):
    """Experiment with different hyperparameters"""
    
    print("\n" + "="*80)
    print("HYPERPARAMETER TUNING EXPERIMENTS")
    print("="*80)
    
    tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
    vocab_size = len(tokenizer)
    
    # Create simple dataset
    print("\nPreparing dataset...")
    try:
        dataset = load_dataset('imdb', split='train[:2000]')
        train_data = dataset.train_test_split(test_size=0.2, seed=42)
        train_texts = train_data['train']['text']
        train_labels = train_data['train']['label']
        val_texts = train_data['test']['text']
        val_labels = train_data['test']['label']
    except:
        train_texts = ["This movie is great!"] * 100 + ["This movie is terrible!"] * 100
        train_labels = [1] * 100 + [0] * 100
        val_texts = ["Amazing film!"] * 20 + ["Awful movie!"] * 20
        val_labels = [1] * 20 + [0] * 20
    
    train_dataset = SentimentDataset(train_texts, train_labels, tokenizer, max_len=128)
    val_dataset = SentimentDataset(val_texts, val_labels, tokenizer, max_len=128)
    
    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=16)
    
    # Experiment configurations
    configs = [
        {'name': 'Tiny', 'd_model': 64, 'num_layers': 2, 'num_heads': 2, 'd_ff': 256},
        {'name': 'Small', 'd_model': 128, 'num_layers': 2, 'num_heads': 4, 'd_ff': 512},
        {'name': 'Medium', 'd_model': 256, 'num_layers': 4, 'num_heads': 8, 'd_ff': 1024},
        {'name': 'Large', 'd_model': 384, 'num_layers': 6, 'num_heads': 8, 'd_ff': 1536},
    ]
    
    results = {}
    
    for config in configs:
        print(f"\n{'-'*80}")
        print(f"Testing Configuration: {config['name']}")
        print(f"  d_model={config['d_model']}, layers={config['num_layers']}, "
              f"heads={config['num_heads']}, d_ff={config['d_ff']}")
        print(f"{'-'*80}")
        
        bert = BERT(
            vocab_size=vocab_size,
            d_model=config['d_model'],
            num_layers=config['num_layers'],
            num_heads=config['num_heads'],
            d_ff=config['d_ff'],
            max_len=128,
            dropout=0.1
        )
        
        model = CustomBERTForSequenceClassification(bert, num_classes=2)
        param_count = count_parameters(model)
        print(f"Parameters: {param_count:,}")
        
        history = train_sentiment_model(
            model, train_loader, val_loader, epochs=3, lr=2e-4, 
            device=device, model_name=config['name']
        )
        
        history['param_count'] = param_count
        results[config['name']] = history
    
    # Visualize hyperparameter impact
    visualize_hyperparameter_results(results)
    
    return results


def visualize_hyperparameter_results(results):
    """Visualize hyperparameter tuning results"""
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('Hyperparameter Tuning Results', fontsize=16, fontweight='bold')
    
    configs = list(results.keys())
    colors = plt.cm.viridis(np.linspace(0, 1, len(configs)))
    
    # 1. Accuracy vs Model Size
    params = [results[c]['param_count'] for c in configs]
    accs = [results[c]['best_val_acc'] for c in configs]
    
    axes[0, 0].scatter(params, accs, s=200, c=colors, alpha=0.6, edgecolors='black', linewidth=2)
    for i, config in enumerate(configs):
        axes[0, 0].annotate(config, (params[i], accs[i]), 
                           xytext=(10, 5), textcoords='offset points', fontsize=9)
    axes[0, 0].set_xlabel('Number of Parameters')
    axes[0, 0].set_ylabel('Best Validation Accuracy')
    axes[0, 0].set_title('Accuracy vs Model Size')
    axes[0, 0].grid(alpha=0.3)
    
    # 2. Training Time vs Model Size
    train_times = [sum(results[c]['train_time']) for c in configs]
    
    axes[0, 1].scatter(params, train_times, s=200, c=colors, alpha=0.6, edgecolors='black', linewidth=2)
    for i, config in enumerate(configs):
        axes[0, 1].annotate(config, (params[i], train_times[i]), 
                           xytext=(10, 5), textcoords='offset points', fontsize=9)
    axes[0, 1].set_xlabel('Number of Parameters')
    axes[0, 1].set_ylabel('Total Training Time (seconds)')
    axes[0, 1].set_title('Training Time vs Model Size')
    axes[0, 1].grid(alpha=0.3)
    
    # 3. Learning Curves Comparison
    for i, config in enumerate(configs):
        epochs = range(1, len(results[config]['val_acc']) + 1)
        axes[1, 0].plot(epochs, results[config]['val_acc'], 
                       marker='o', label=config, color=colors[i], linewidth=2)
    axes[1, 0].set_xlabel('Epoch')
    axes[1, 0].set_ylabel('Validation Accuracy')
    axes[1, 0].set_title('Learning Curves by Configuration')
    axes[1, 0].legend()
    axes[1, 0].grid(alpha=0.3)
    
    # 4. Efficiency: Accuracy per Parameter
    efficiency = [accs[i] / (params[i] / 1e6) for i in range(len(configs))]  # Acc per million params
    
    axes[1, 1].bar(range(len(configs)), efficiency, color=colors, alpha=0.7, edgecolor='black')
    axes[1, 1].set_xticks(range(len(configs)))
    axes[1, 1].set_xticklabels(configs, rotation=45)
    axes[1, 1].set_ylabel('Accuracy per Million Parameters')
    axes[1, 1].set_title('Parameter Efficiency')
    axes[1, 1].grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('hyperparameter_tuning_results.png', dpi=300, bbox_inches='tight')
    print("\n✓ Hyperparameter results saved to 'hyperparameter_tuning_results.png'")
    plt.show()


# MAIN EXECUTION


if __name__ == "__main__":
    # Set device
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"\n{'='*80}")
    print(f"Using device: {device}")
    if device.type == 'cuda':
        print(f"GPU: {torch.cuda.get_device_name(0)}")
        print(f"Memory Available: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
    print(f"{'='*80}\n")
    
    # Run experiments
    print("Starting BERT comparison experiments...")
    print("This will compare custom BERT models with different configurations")
    print("against pre-trained BERT-base model.\n")
    
    # 1. Sentiment Analysis Comparison
    sentiment_results = run_sentiment_comparison(device=device)
    
    # 2. Question Answering Comparison (optional, comment out if too slow)
    # qa_results = run_qa_comparison(device=device)
    qa_results = None
    
    # 3. Hyperparameter Tuning Experiments
    hyperparam_results = hyperparameter_experiment(device=device)
    
    # 4. Visualize and print results
    print("\n" + "="*80)
    print("GENERATING VISUALIZATIONS AND SUMMARY")
    print("="*80)
    
    visualize_results(sentiment_results, qa_results)
    print_detailed_comparison(sentiment_results, qa_results)
    
    # 5. Save results to file
    print("\nSaving results to CSV...")
    summary_data = []
    for model_name, metrics in sentiment_results.items():
        summary_data.append({
            'Model': model_name,
            'Best_Val_Accuracy': metrics['best_val_acc'],
            'Final_F1_Score': metrics['final_f1'],
            'Avg_Train_Loss': np.mean(metrics['train_loss']),
            'Avg_Val_Loss': np.mean(metrics['val_loss']),
            'Total_Train_Time': sum(metrics['train_time']),
            'Avg_Inference_Time': np.mean(metrics['inference_time'])
        })
    
    df_summary = pd.DataFrame(summary_data)
    df_summary.to_csv('bert_comparison_summary.csv', index=False)
    print("✓ Summary saved to 'bert_comparison_summary.csv'")
    
    print("\n" + "="*80)
    print("EXPERIMENT COMPLETE!")
    print("="*80)
    print("\nGenerated Files:")
    print("  • bert_comparison_results.png - Visual comparison of all models")
    print("  • hyperparameter_tuning_results.png - Hyperparameter analysis")
    print("  • bert_comparison_summary.csv - Detailed metrics table")
    print("\nKey Takeaways:")
    print("  1. Pre-trained BERT shows strong transfer learning capabilities")
    print("  2. Larger custom models approach pre-trained performance but need more training")
    print("  3. Smaller custom models are faster but may plateau at lower accuracy")
    print("  4. Model size should be chosen based on task complexity and data availability")
    print("="*80)
```
