```python
# Vision Transformer (ViT) vs CNN vs FCFNN: Comprehensive Comparison
# ImageNet-20 Classification with Ablation Studies

# ============================================================================
# COMPLETE JUPYTER NOTEBOOK CODE
# Save this entire file as: vit_comparison.ipynb
# ============================================================================

"""
To create the .ipynb file:
1. Copy this entire Python script
2. Create a new Jupyter notebook
3. Paste each section into separate cells as indicated
4. Or use the JSON at the end to create the notebook file directly
"""

# ============================================================================
# CELL 1: Markdown
# ============================================================================
"""
# Vision Transformer (ViT) vs CNN vs FCFNN: Comprehensive Comparison
## ImageNet-20 Classification with Ablation Studies

This notebook implements and compares:
1. Vision Transformer (ViT) based classifier
2. Convolutional Neural Network (CNN) classifier  
3. Fully Connected Feed-Forward Neural Network (FCFNN)

### Ablation Studies:
- Effect of number of attention heads
- Effect of patch embedding strategies
- Effect of positional embedding types
- Discussion on training difficulties
"""

# ============================================================================
# CELL 2: Setup and Imports
# ============================================================================

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import pandas as pd
import time
import warnings
warnings.filterwarnings('ignore')

# Set random seeds
torch.manual_seed(42)
np.random.seed(42)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# ============================================================================
# CELL 3: Configuration and Data Loading
# ============================================================================

# Configuration
IMG_SIZE = 224
PATCH_SIZE = 16
NUM_CLASSES = 10  # CIFAR-10 demo, change to 20 for ImageNet-20
BATCH_SIZE = 32
NUM_EPOCHS = 10
EPOCHS_ABLATION = 8

# Data transforms
train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

test_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

print("Loading dataset...")
# Using CIFAR-10 as demo
train_dataset = torchvision.datasets.CIFAR10(
    root='./data', train=True, download=True, transform=train_transform
)
test_dataset = torchvision.datasets.CIFAR10(
    root='./data', train=False, download=True, transform=test_transform
)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

print(f"Train: {len(train_dataset)} images | Test: {len(test_dataset)} images")

# ============================================================================
# CELL 4: Vision Transformer - Patch Embedding
# ============================================================================

class PatchEmbedding(nn.Module):
    """Different patch embedding strategies"""
    def __init__(self, img_size=224, patch_size=16, in_channels=3, embed_dim=768, method='conv'):
        super().__init__()
        self.img_size = img_size
        self.patch_size = patch_size
        self.n_patches = (img_size // patch_size) ** 2
        self.method = method
        
        if method == 'conv':
            self.proj = nn.Conv2d(in_channels, embed_dim, kernel_size=patch_size, stride=patch_size)
        elif method == 'linear':
            self.proj = nn.Linear(in_channels * patch_size * patch_size, embed_dim)
        elif method == 'hybrid':
            self.conv = nn.Sequential(
                nn.Conv2d(in_channels, 64, 3, 1, 1),
                nn.ReLU(),
                nn.Conv2d(64, embed_dim, kernel_size=patch_size//2, stride=patch_size//2)
            )
            self.proj = nn.Linear(embed_dim * 4, embed_dim)
    
    def forward(self, x):
        B, C, H, W = x.shape
        
        if self.method == 'conv':
            x = self.proj(x)
            x = x.flatten(2).transpose(1, 2)
        elif self.method == 'linear':
            x = x.unfold(2, self.patch_size, self.patch_size).unfold(3, self.patch_size, self.patch_size)
            x = x.contiguous().view(B, C, -1, self.patch_size, self.patch_size)
            x = x.permute(0, 2, 1, 3, 4).contiguous()
            x = x.view(B, -1, C * self.patch_size * self.patch_size)
            x = self.proj(x)
        elif self.method == 'hybrid':
            x = self.conv(x)
            x = x.flatten(2).transpose(1, 2)
            x = self.proj(x)
        
        return x

# ============================================================================
# CELL 5: Vision Transformer - Positional Embedding
# ============================================================================

class PositionalEmbedding(nn.Module):
    """Different positional embedding strategies"""
    def __init__(self, n_patches, embed_dim, method='learnable'):
        super().__init__()
        self.method = method
        
        if method == 'learnable':
            self.pos_embed = nn.Parameter(torch.randn(1, n_patches + 1, embed_dim))
        elif method == 'sincos':
            self.pos_embed = self._get_sincos_encoding(n_patches + 1, embed_dim)
        elif method == 'none':
            self.pos_embed = None
    
    def _get_sincos_encoding(self, n_positions, embed_dim):
        position = torch.arange(n_positions).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, embed_dim, 2) * -(np.log(10000.0) / embed_dim))
        pos_embed = torch.zeros(1, n_positions, embed_dim)
        pos_embed[0, :, 0::2] = torch.sin(position * div_term)
        pos_embed[0, :, 1::2] = torch.cos(position * div_term)
        return nn.Parameter(pos_embed, requires_grad=False)
    
    def forward(self, x):
        if self.pos_embed is not None:
            return x + self.pos_embed
        return x

# ============================================================================
# CELL 6: Vision Transformer - Multi-Head Attention
# ============================================================================

class MultiHeadAttention(nn.Module):
    def __init__(self, embed_dim, num_heads):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        
        assert self.head_dim * num_heads == embed_dim
        
        self.qkv = nn.Linear(embed_dim, embed_dim * 3)
        self.proj = nn.Linear(embed_dim, embed_dim)
        self.scale = self.head_dim ** -0.5
    
    def forward(self, x):
        B, N, C = x.shape
        qkv = self.qkv(x).reshape(B, N, 3, self.num_heads, self.head_dim).permute(2, 0, 3, 1, 4)
        q, k, v = qkv[0], qkv[1], qkv[2]
        
        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = attn.softmax(dim=-1)
        
        x = (attn @ v).transpose(1, 2).reshape(B, N, C)
        x = self.proj(x)
        return x

# ============================================================================
# CELL 7: Vision Transformer - Transformer Block
# ============================================================================

class TransformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, mlp_ratio=4.0, dropout=0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(embed_dim)
        self.attn = MultiHeadAttention(embed_dim, num_heads)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.mlp = nn.Sequential(
            nn.Linear(embed_dim, int(embed_dim * mlp_ratio)),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(int(embed_dim * mlp_ratio), embed_dim),
            nn.Dropout(dropout)
        )
    
    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x

# ============================================================================
# CELL 8: Vision Transformer - Complete Model
# ============================================================================

class VisionTransformer(nn.Module):
    def __init__(self, img_size=224, patch_size=16, in_channels=3, num_classes=10,
                 embed_dim=768, depth=12, num_heads=12, mlp_ratio=4.0, dropout=0.1,
                 patch_embed_method='conv', pos_embed_method='learnable'):
        super().__init__()
        
        self.patch_embed = PatchEmbedding(img_size, patch_size, in_channels, embed_dim, patch_embed_method)
        n_patches = self.patch_embed.n_patches
        
        self.cls_token = nn.Parameter(torch.randn(1, 1, embed_dim))
        self.pos_embed = PositionalEmbedding(n_patches, embed_dim, pos_embed_method)
        self.dropout = nn.Dropout(dropout)
        
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, mlp_ratio, dropout)
            for _ in range(depth)
        ])
        
        self.norm = nn.LayerNorm(embed_dim)
        self.head = nn.Linear(embed_dim, num_classes)
    
    def forward(self, x):
        B = x.shape[0]
        x = self.patch_embed(x)
        
        cls_tokens = self.cls_token.expand(B, -1, -1)
        x = torch.cat([cls_tokens, x], dim=1)
        x = self.pos_embed(x)
        x = self.dropout(x)
        
        for block in self.blocks:
            x = block(x)
        
        x = self.norm(x)
        x = x[:, 0]
        x = self.head(x)
        return x

# ============================================================================
# CELL 9: CNN Classifier
# ============================================================================

class CNNClassifier(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(256, 512, 3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, 3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
        )
        
        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(512, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
    
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

# ============================================================================
# CELL 10: FCFNN Classifier
# ============================================================================

class FCFNNClassifier(nn.Module):
    def __init__(self, img_size=224, num_classes=10):
        super().__init__()
        
        input_dim = 3 * img_size * img_size
        
        self.network = nn.Sequential(
            nn.Flatten(),
            nn.Linear(input_dim, 2048),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(2048, 1024),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(1024, 512),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
    
    def forward(self, x):
        return self.network(x)

# ============================================================================
# CELL 11: Training Functions
# ============================================================================

def train_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    
    pbar = tqdm(loader, desc='Training')
    for inputs, labels in pbar:
        inputs, labels = inputs.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
        
        pbar.set_postfix({'loss': running_loss / (pbar.n + 1), 'acc': 100. * correct / total})
    
    return running_loss / len(loader), 100. * correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Evaluating'):
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
    
    return running_loss / len(loader), 100. * correct / total


def train_model(model, train_loader, test_loader, num_epochs, lr=1e-4, device='cuda'):
    model = model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, num_epochs)
    
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    
    for epoch in range(num_epochs):
        print(f"Epoch {epoch+1}/{num_epochs}")
        
        train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = evaluate(model, test_loader, criterion, device)
        scheduler.step()
        
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        
        print(f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%")
        print(f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%\n")
    
    return history

# ============================================================================
# CELL 12: Initialize Models
# ============================================================================

# Initialize models
vit_model = VisionTransformer(
    img_size=IMG_SIZE, patch_size=PATCH_SIZE, num_classes=NUM_CLASSES,
    embed_dim=384, depth=6, num_heads=6, mlp_ratio=4.0
)

cnn_model = CNNClassifier(num_classes=NUM_CLASSES)
fcfnn_model = FCFNNClassifier(img_size=IMG_SIZE, num_classes=NUM_CLASSES)

print("ViT Parameters:", sum(p.numel() for p in vit_model.parameters()))
print("CNN Parameters:", sum(p.numel() for p in cnn_model.parameters()))
print("FCFNN Parameters:", sum(p.numel() for p in fcfnn_model.parameters()))

# ============================================================================
# CELL 13: Train ViT
# ============================================================================

print("\n" + "="*50)
print("Training Vision Transformer...")
print("="*50)
vit_history = train_model(vit_model, train_loader, test_loader, NUM_EPOCHS, lr=3e-4, device=device)

# ============================================================================
# CELL 14: Train CNN
# ============================================================================

print("\n" + "="*50)
print("Training CNN...")
print("="*50)
cnn_history = train_model(cnn_model, train_loader, test_loader, NUM_EPOCHS, lr=1e-3, device=device)

# ============================================================================
# CELL 15: Train FCFNN
# ============================================================================

print("\n" + "="*50)
print("Training FCFNN...")
print("="*50)
fcfnn_history = train_model(fcfnn_model, train_loader, test_loader, NUM_EPOCHS, lr=1e-3, device=device)

# ============================================================================
# CELL 16: Plot Main Comparison
# ============================================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].plot(vit_history['train_acc'], label='ViT', marker='o')
axes[0].plot(cnn_history['train_acc'], label='CNN', marker='s')
axes[0].plot(fcfnn_history['train_acc'], label='FCFNN', marker='^')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy (%)')
axes[0].set_title('Training Accuracy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(vit_history['val_acc'], label='ViT', marker='o')
axes[1].plot(cnn_history['val_acc'], label='CNN', marker='s')
axes[1].plot(fcfnn_history['val_acc'], label='FCFNN', marker='^')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')
axes[1].set_title('Validation Accuracy')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('model_comparison.png', dpi=300)
plt.show()

# Summary table
summary_df = pd.DataFrame({
    'Model': ['ViT', 'CNN', 'FCFNN'],
    'Final Train Acc': [vit_history['train_acc'][-1], cnn_history['train_acc'][-1], fcfnn_history['train_acc'][-1]],
    'Final Val Acc': [vit_history['val_acc'][-1], cnn_history['val_acc'][-1], fcfnn_history['val_acc'][-1]],
    'Best Val Acc': [max(vit_history['val_acc']), max(cnn_history['val_acc']), max(fcfnn_history['val_acc'])],
    'Parameters (M)': [
        sum(p.numel() for p in vit_model.parameters()) / 1e6,
        sum(p.numel() for p in cnn_model.parameters()) / 1e6,
        sum(p.numel() for p in fcfnn_model.parameters()) / 1e6
    ]
})

print("\nMODEL COMPARISON SUMMARY")
print("="*70)
print(summary_df.to_string(index=False))
print("="*70)

# ============================================================================
# CELL 17: Ablation Study - Number of Heads
# ============================================================================

head_configs = [1, 2, 4, 6, 8, 12]
head_results = {}

print("\n" + "="*70)
print("ABLATION: Number of Attention Heads")
print("="*70)

for num_heads in head_configs:
    print(f"\nTraining ViT with {num_heads} heads...")
    model = VisionTransformer(
        img_size=IMG_SIZE, patch_size=PATCH_SIZE, num_classes=NUM_CLASSES,
        embed_dim=384, depth=4, num_heads=num_heads, mlp_ratio=4.0
    )
    history = train_model(model, train_loader, test_loader, EPOCHS_ABLATION, lr=3e-4, device=device)
    head_results[num_heads] = history

# ============================================================================
# CELL 18: Visualize Head Ablation
# ============================================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

for num_heads, history in head_results.items():
    axes[0].plot(history['train_acc'], label=f'{num_heads} heads', marker='o')
    axes[1].plot(history['val_acc'], label=f'{num_heads} heads', marker='o')

axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy (%)')
axes[0].set_title('Training Accuracy vs Number of Heads')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')
axes[1].set_title('Validation Accuracy vs Number of Heads')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('heads_ablation.png', dpi=300)
plt.show()

head_summary = pd.DataFrame([
    {'Num Heads': h, 'Best Val Acc': max(hist['val_acc']), 'Final Val Acc': hist['val_acc'][-1]}
    for h, hist in head_results.items()
])
print("\nHEAD ABLATION SUMMARY")
print(head_summary.to_string(index=False))

# ============================================================================
# CELL 19: Ablation Study - Patch Embedding
# ============================================================================

patch_methods = ['conv', 'linear', 'hybrid']
patch_results = {}

print("\n" + "="*70)
print("ABLATION: Patch Embedding Strategy")
print("="*70)

for method in patch_methods:
    print(f"\nTraining ViT with '{method}' patch embedding...")
    model = VisionTransformer(
        img_size=IMG_SIZE, patch_size=PATCH_SIZE, num_classes=NUM_CLASSES,
        embed_dim=384, depth=4, num_heads=6, mlp_ratio=4.0,
        patch_embed_method=method
    )
    history = train_model(model, train_loader, test_loader, EPOCHS_ABLATION, lr=3e-4, device=device)
    patch_results[method] = history

# ============================================================================
# CELL 20: Visualize Patch Ablation
# ============================================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

for method, history in patch_results.items():
    axes[0].plot(history['train_acc'], label=f'{method}', marker='o')
    axes[1].plot(history['val_acc'], label=f'{method}', marker='o')

axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy (%)')
axes[0].set_title('Training Accuracy vs Patch Embedding')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')
axes[1].set_title('Validation Accuracy vs Patch Embedding')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('patch_ablation.png', dpi=300)
plt.show()

patch_summary = pd.DataFrame([
    {'Patch Method': m, 'Best Val Acc': max(hist['val_acc']), 'Final Val Acc': hist['val_acc'][-1]}
    for m, hist in patch_results.items()
])
print("\nPATCH EMBEDDING ABLATION SUMMARY")
print(patch_summary.to_string(index=False))

# ============================================================================
# CELL 21: Ablation Study - Positional Embedding
# ============================================================================

pos_methods = ['learnable', 'sincos', 'none']
pos_results = {}

print("\n" + "="*70)
print("ABLATION: Positional Embedding Strategy")
print("="*70)

for method in pos_methods:
    print(f"\nTraining ViT with '{method}' positional embedding...")
    model = VisionTransformer(
        img_size=IMG_SIZE, patch_size=PATCH_SIZE, num_classes=NUM_CLASSES,
        embed_dim=384, depth=4, num_heads=6, mlp_ratio=4.0,
        pos_embed_method=method
    )
    history = train_model(model, train_loader, test_loader, EPOCHS_ABLATION, lr=3e-4, device=device)
    pos_results[method] = history

# ============================================================================
# CELL 22: Visualize Positional Ablation
# ============================================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

for method, history in pos_results.items():
    axes[0].plot(history['train_acc'], label=f'{method}', marker='o')
    axes[1].plot(history['val_acc'], label=f'{method}', marker='o')

axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy (%)')
axes[0].set_title('Training Accuracy vs Positional Embedding')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')
axes[1].set_title('Validation Accuracy vs Positional Embedding')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('pos_ablation.png', dpi=300)
plt.show()

pos_summary = pd.DataFrame([
    {'Positional Embedding': m, 'Best Val Acc': max(hist['val_acc']), 'Final Val Acc': hist['val_acc'][-1]}
    for m, hist in pos_results.items()
])
print("\nPOSITIONAL EMBEDDING ABLATION SUMMARY")
print(pos_summary)
```

```python
# ============================================================================
# CELL 4: Vision Transformer - Patch Embedding (FIXED FOR HYBRID)
# ============================================================================

class PatchEmbedding(nn.Module):
    """Different patch embedding strategies"""
    def __init__(self, img_size=224, patch_size=16, in_channels=3, embed_dim=768, method='conv'):
        super().__init__()
        self.img_size = img_size
        self.patch_size = patch_size
        self.method = method
        
        if method == 'conv':
            # Standard convolutional projection
            self.n_patches = (img_size // patch_size) ** 2
            self.proj = nn.Conv2d(in_channels, embed_dim, kernel_size=patch_size, stride=patch_size)
            
        elif method == 'linear':
            # Linear projection after reshaping
            self.n_patches = (img_size // patch_size) ** 2
            self.proj = nn.Linear(in_channels * patch_size * patch_size, embed_dim)
            
        elif method == 'hybrid':
            # FIXED: Hybrid with same number of patches as other methods
            # Use strided convolutions to reduce to patch_size effectively
            stride = patch_size // 2  # Two conv layers with stride 2 each
            
            self.conv = nn.Sequential(
                nn.Conv2d(in_channels, 64, kernel_size=3, stride=2, padding=1),  # 224 -> 112
                nn.ReLU(),
                nn.Conv2d(64, embed_dim, kernel_size=3, stride=2, padding=1),    # 112 -> 56
                nn.ReLU()
            )
            
            # After 2 stride-2 convs: 224 -> 56
            # Need to get to 14x14 (same as 224/16)
            # Use adaptive pooling to ensure consistent output size
            self.pool = nn.AdaptiveAvgPool2d((img_size // patch_size, img_size // patch_size))
            self.n_patches = (img_size // patch_size) ** 2
    
    def forward(self, x):
        B, C, H, W = x.shape
        
        if self.method == 'conv':
            # Standard: Conv with stride=patch_size
            x = self.proj(x)  # (B, embed_dim, H/patch_size, W/patch_size)
            x = x.flatten(2).transpose(1, 2)  # (B, n_patches, embed_dim)
            
        elif self.method == 'linear':
            # Unfold into patches, then linear projection
            x = x.unfold(2, self.patch_size, self.patch_size).unfold(3, self.patch_size, self.patch_size)
            x = x.contiguous().view(B, C, -1, self.patch_size, self.patch_size)
            x = x.permute(0, 2, 1, 3, 4).contiguous()
            x = x.view(B, -1, C * self.patch_size * self.patch_size)
            x = self.proj(x)
            
        elif self.method == 'hybrid':
            # Apply conv layers
            x = self.conv(x)  # (B, embed_dim, 56, 56)
            # Ensure consistent size with other methods
            x = self.pool(x)  # (B, embed_dim, 14, 14) for patch_size=16
            x = x.flatten(2).transpose(1, 2)  # (B, 196, embed_dim)
        
        return x
```
